## Gender Distribution Across Topics Among Sanctioned Individuals

### Analytical Objective

The analysis investigates whether the distribution of thematic topics differs between female and male individuals within the population of sanctioned persons. The goal is to assess **associations between gender and topic classification**, without making causal claims.

---

### Data Structure and Method

A contingency table was constructed reporting, for each topic, the number of female and male individuals associated with that topic. The analysis proceeded in three stages:

1. **Global association test**
   - A chi-square test of independence evaluated whether gender and topic assignment are statistically associated overall.

2. **Topic-level comparisons**
   - For each topic, a 2×2 contingency table was constructed comparing:
     - presence vs absence of the topic
     - female vs male individuals.

3. **Statistical estimation**
   - Fisher’s exact test was used to compute:
     - odds ratios (effect size),
     - 95% confidence intervals,
     - p-values.
   - Multiple testing was controlled using **False Discovery Rate (Benjamini–Hochberg correction)**.

---

### Global Result

The chi-square test indicates a strong association between gender and topic distribution:

\[
\chi^2(17) = 506.41,\quad p < 0.001
\]

This result shows that gender proportions are **not uniformly distributed across topics**.

---

### Interpretation of Effect Sizes

The odds ratio (OR) measures relative representation:

- **OR = 1** → no gender difference  
- **OR > 1** → topic relatively more associated with females  
- **OR < 1** → topic relatively more associated with males  

Confidence intervals excluding 1 indicate statistically reliable effects after correction.

---

### Topics Overrepresented Among Females

Several topics show significantly higher odds for females:

- **role.rca** (OR ≈ 3.86) — strongest observed association
- **role.judge** (OR ≈ 2.59)
- **sanction.counter** (OR ≈ 1.84)
- **poi** (OR ≈ 1.54)
- **role.pep** (OR ≈ 1.25)
- **corp.disqual** (small but significant effect)

These categories are predominantly linked to institutional, regulatory, or administrative roles.

---

### Topics Overrepresented Among Males

Strong male-associated topics include:

- **mil** (OR ≈ 0.12)
- **crime.war** (OR ≈ 0.09)
- **role.oligarch**
- **crime.terror**
- **export.control**
- **crime**
- **wanted**

These topics largely correspond to military, armed conflict, or severe criminal contexts.

---

### Topics Without Detectable Gender Differences

Some topics do not show statistically meaningful differences:

- **sanction**
- **role.diplo**
- **reg.action**
- **asset.frozen** (very sparse observations)

For these categories, gender proportions align with the overall dataset baseline.

---

### Sparse Categories

Topics with zero observations in one gender produce undefined or infinite log-odds values. These cases reflect **insufficient data**, not extreme effects, and should be interpreted cautiously.

---

### Substantive Interpretation

The results indicate that gender differences among sanctioned individuals are **structurally heterogeneous** rather than uniform. The observed variation appears concentrated along a thematic distinction:

- institutional and regulatory roles show relatively higher female representation;
- military and conflict-related domains show strong male predominance.

Thus, the global statistical association is driven by specific topic domains rather than a general gender imbalance across all sanction contexts.

---

### Limitations

- The analysis identifies statistical association, not causation.
- Results depend on topic classification practices within the dataset.
- Individuals may belong to multiple topics, introducing dependence between observations.
- Sparse categories reduce reliability of some estimates.

---

### Conclusion

Gender distribution varies significantly across sanction-related topics. Topic-level analysis reveals systematic differences, with female overrepresentation in institutional roles and male overrepresentation in conflict and military domains. These findings highlight the importance of examining thematic structure when interpreting demographic patterns within sanctions data.

## Gender × Sanctioned Entity Type: Statistical Analysis

### Data & Design

The dataset contains **257 sanctioned individuals** (85 female, 172 male) classified
across five non-mutually exclusive entity types:

| Code | Type label |
|------|------------|
| C | Activity-based ties |
| D | Profit-based ties |
| E | Status-based ties |
| F | Family ties |
| G | Other ties |

Because individuals can hold multiple types simultaneously, each type was treated as
an independent binary indicator. Per-type associations were tested with **Fisher's
exact test** (two-sided), using the true per-gender individual counts as marginals.
Raw p-values were adjusted for multiple comparisons with the
**Benjamini–Hochberg false discovery rate** procedure (α = 0.05).
Effect sizes are reported as **odds ratios (OR)** with 95% confidence intervals;
log-odds are included for symmetry of interpretation.

---

### Global Association

A Pearson chi-square test on the full Type × Gender count matrix
yielded **χ²(4) = 57.08, p = 1.19 × 10⁻¹¹**, indicating strong overall dependence
between gender and type distribution. Note that because individuals can appear in
multiple type rows, the independence assumption is not fully met; this statistic
should be read as descriptive of the aggregate pattern rather than a formal test.

---

### Per-Type Results

| Type | Female *n* | Male *n* | OR | 95% CI | *p* (raw) | *p* (BH-adj) | log-OR |
|------|:---:|:---:|:---:|---|:---:|:---:|:---:|
| F — Family ties | 52 | 35 | **4.12** | [2.45, 6.92] | 8.57 × 10⁻⁸ | 1.43 × 10⁻⁷ | +1.41 |
| E — Status ties | 83 | 174 | 0.48 | [0.27, 0.84] | 1.13 × 10⁻² | 1.42 × 10⁻² | −0.74 |
| C — Activity ties | 40 | 153 | 0.18 | [0.11, 0.30] | 7.08 × 10⁻¹² | 1.77 × 10⁻¹¹ | −1.70 |
| G — Other ties | 35 | 151 | 0.16 | [0.09, 0.26] | 1.15 × 10⁻¹³ | 5.76 × 10⁻¹³ | −1.85 |
| D — Profit ties | 19 | 46 | 0.69 | [0.38, 1.26] | 2.48 × 10⁻¹ | 2.48 × 10⁻¹ | −0.36 |

Sorted by |log-OR| descending. OR > 1 = female-skewed; OR < 1 = male-skewed.

**Standardized Pearson residuals** confirm the pattern: the Family ties cell drives
the signal most strongly (Female: +5.31, Male: −3.40), followed by Other ties
(Female: −2.59) and Activity ties (Female: −2.15). Status and Profit ties show
residuals close to zero.

---

### Composite-Condition Results

Four theoretically motivated composite conditions were tested separately,
again with Fisher's exact test and BH correction across the four comparisons.

| Condition | Female *n* | Male *n* | OR | 95% CI | *p* (raw) | *p* (BH-adj) | log-OR |
|---|:---:|:---:|:---:|---|:---:|:---:|:---:|
| Family ties only (F) | 52 | 35 | **4.12** | [2.45, 6.92] | 8.57 × 10⁻⁸ | 1.71 × 10⁻⁷ | +1.41 |
| Profit + Family (D ∩ F) | 16 | 20 | 1.52 | [0.75, 3.06] | 2.69 × 10⁻¹ | 3.59 × 10⁻¹ | +0.42 |
| Activity + Family (C ∩ F) | 9 | 21 | 0.75 | [0.33, 1.71] | 5.54 × 10⁻¹ | 5.54 × 10⁻¹ | −0.28 |
| Activity + Other (C ∩ G) | 35 | 144 | **0.19** | [0.11, 0.31] | 9.21 × 10⁻¹² | 3.68 × 10⁻¹¹ | −1.68 |

The two composite conditions that condition on an additional type beyond Family ties
(D ∩ F, C ∩ F) lose significance entirely, likely reflecting small cell sizes
(n = 16 and n = 9 females respectively) rather than a true null effect.

---

### Summary

Four of five types show statistically significant gender skew after FDR correction:

- **Female-skewed:** Family ties (OR = 4.12) — females are ~4× more likely than
  males to be sanctioned on the basis of family relationships.
- **Male-skewed:** Other ties (OR = 0.16) and Activity ties (OR = 0.18) show the
  strongest male concentration (~5–6× more likely); Status ties show a weaker but
  significant male skew (OR = 0.48).
- **No significant difference:** Profit-based ties (OR = 0.69, p = 0.25).

The Activity + Other composite (C ∩ G) replicates the male skew at high significance
(OR = 0.19), consistent with the marginal effects of both constituent types.

---

> **Interpretive convention:** ORs are framed as "odds of a female individual
> belonging to this type relative to a male individual." To reverse the reference
> category, take the reciprocal of each OR and negate each log-OR; CIs invert
> accordingly.
